```python
"""
Phase 1: Indiana University CXR Dataset - Data Loading and Preprocessing
Dataset: Indiana University Chest X-Ray Collection
"""

import os
import pandas as pd
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path
import re

# ============================================
# INDIANA DATASET LOADER
# ============================================

class IndianaDatasetLoader:
    """
    Loads and preprocesses Indiana University Chest X-ray dataset

    Dataset structure:
    - images/ (folder with CXR images)
    - indiana_projections.csv (image metadata)
    - indiana_reports.csv (radiology reports)
    """

    def __init__(self, dataset_path):
        """
        Args:
            dataset_path: Path to the root directory of Indiana dataset
        """
        self.dataset_path = Path(dataset_path)
        self.images_path = self.dataset_path / "images"
        self.projections_csv = self.dataset_path / "indiana_projections.csv"
        self.reports_csv = self.dataset_path / "indiana_reports.csv"

        # Load CSV files
        self.projections_df = None
        self.reports_df = None
        self.merged_df = None

        # CheXpert-style labels (we'll extract from reports)
        self.chexpert_labels = [
            'No Finding', 'Enlarged Cardiomediastinum', 'Cardiomegaly',
            'Lung Opacity', 'Lung Lesion', 'Edema', 'Consolidation',
            'Pneumonia', 'Atelectasis', 'Pneumothorax', 'Pleural Effusion',
            'Pleural Other', 'Fracture', 'Support Devices'
        ]

    def load_data(self):
        """Load projection and report CSV files"""
        print("Loading Indiana dataset...")

        # Load projections
        if self.projections_csv.exists():
            self.projections_df = pd.read_csv(self.projections_csv)
            print(f"✓ Loaded projections: {len(self.projections_df)} records")
            print(f"  Columns: {list(self.projections_df.columns)}")
        else:
            raise FileNotFoundError(f"Projections file not found: {self.projections_csv}")

        # Load reports
        if self.reports_csv.exists():
            self.reports_df = pd.read_csv(self.reports_csv)
            print(f"✓ Loaded reports: {len(self.reports_df)} records")
            print(f"  Columns: {list(self.reports_df.columns)}")
        else:
            raise FileNotFoundError(f"Reports file not found: {self.reports_csv}")

        return self.projections_df, self.reports_df

    def merge_data(self):
        """Merge projections and reports based on patient/image ID"""
        if self.projections_df is None or self.reports_df is None:
            raise ValueError("Load data first using load_data()")

        # Merge on common identifier (usually 'uid' or 'filename')
        # Adjust the merge key based on your CSV structure
        common_cols = set(self.projections_df.columns) & set(self.reports_df.columns)
        print(f"\nCommon columns for merging: {common_cols}")

        # Typical merge on 'uid' or 'filename'
        if 'uid' in common_cols:
            self.merged_df = pd.merge(
                self.projections_df,
                self.reports_df,
                on='uid',
                how='inner'
            )
        elif 'filename' in common_cols:
            self.merged_df = pd.merge(
                self.projections_df,
                self.reports_df,
                on='filename',
                how='inner'
            )
        else:
            print("Warning: No obvious merge key found. Manual merging needed.")
            return None

        print(f"✓ Merged dataset: {len(self.merged_df)} records")
        return self.merged_df

    def filter_frontal_images(self):
        """Keep only frontal (PA/AP) images, remove lateral"""
        if self.merged_df is None:
            raise ValueError("Merge data first using merge_data()")

        # Filter for frontal projections
        # Typical column name: 'projection' with values 'Frontal', 'PA', 'AP'
        if 'projection' in self.merged_df.columns:
            frontal_keywords = ['frontal', 'pa', 'ap']
            mask = self.merged_df['projection'].str.lower().str.contains(
                '|'.join(frontal_keywords),
                na=False
            )
            self.merged_df = self.merged_df[mask].reset_index(drop=True)
            print(f"✓ Filtered frontal images: {len(self.merged_df)} records")
        else:
            print("Warning: 'projection' column not found. Keeping all images.")

        return self.merged_df

    def extract_labels_from_reports(self, report_column='findings'):
        """
        Extract CheXpert-style labels from radiology reports using keyword matching

        Args:
            report_column: Name of column containing radiology findings/impressions
        """
        if self.merged_df is None:
            raise ValueError("Merge and filter data first")

        print("\nExtracting labels from reports...")

        # Keywords for each finding (case-insensitive)
        keywords_dict = {
            'Cardiomegaly': ['cardiomegaly', 'cardiac enlargement', 'enlarged heart'],
            'Edema': ['edema', 'pulmonary edema', 'fluid overload'],
            'Consolidation': ['consolidation', 'consolidative'],
            'Pneumonia': ['pneumonia', 'infiltrate', 'infection'],
            'Atelectasis': ['atelectasis', 'collapse'],
            'Pneumothorax': ['pneumothorax', 'ptx'],
            'Pleural Effusion': ['pleural effusion', 'effusion'],
            'Lung Opacity': ['opacity', 'opacification', 'opacities'],
            'Lung Lesion': ['lesion', 'nodule', 'mass'],
            'Fracture': ['fracture', 'broken rib'],
            'Enlarged Cardiomediastinum': ['widened mediastinum', 'mediastinal widening'],
            'Support Devices': ['tube', 'catheter', 'device', 'pacemaker', 'wire'],
            'Pleural Other': ['pleural thickening', 'pleural abnormality']
        }

        # Initialize label columns
        for label in self.chexpert_labels:
            self.merged_df[label] = 0

        # Extract labels
        for idx, row in self.merged_df.iterrows():
            report_text = str(row.get(report_column, '')).lower()

            # Check for "normal" or "no finding"
            if any(word in report_text for word in ['normal', 'clear', 'no acute', 'unremarkable']):
                if not any(keyword in report_text for keywords in keywords_dict.values() for keyword in keywords):
                    self.merged_df.at[idx, 'No Finding'] = 1
                    continue

            # Check for each finding
            for label, keywords in keywords_dict.items():
                if any(keyword in report_text for keyword in keywords):
                    self.merged_df.at[idx, label] = 1

        # Statistics
        print("\nLabel Distribution:")
        for label in self.chexpert_labels:
            count = self.merged_df[label].sum()
            percentage = (count / len(self.merged_df)) * 100
            print(f"  {label}: {count} ({percentage:.1f}%)")

        return self.merged_df

    def get_image_path(self, filename):
        """Get full path to image file"""
        return self.images_path / filename

    def load_image(self, filename, target_size=(224, 224)):
        """
        Load and preprocess a single image

        Args:
            filename: Image filename
            target_size: Target size for resizing (default: 224x224 for CNN)

        Returns:
            Preprocessed image as numpy array
        """
        img_path = self.get_image_path(filename)

        if not img_path.exists():
            raise FileNotFoundError(f"Image not found: {img_path}")

        # Load image
        img = Image.open(img_path).convert('RGB')

        # Resize
        img = img.resize(target_size, Image.LANCZOS)

        # Convert to numpy array and normalize
        img_array = np.array(img) / 255.0

        return img_array

    def visualize_sample(self, num_samples=3):
        """Visualize random samples with their labels"""
        if self.merged_df is None:
            raise ValueError("Process data first")

        samples = self.merged_df.sample(n=min(num_samples, len(self.merged_df)))

        fig, axes = plt.subplots(1, num_samples, figsize=(15, 5))
        if num_samples == 1:
            axes = [axes]

        for idx, (ax, (_, row)) in enumerate(zip(axes, samples.iterrows())):
            # Get filename (adjust column name if needed)
            filename = row.get('filename', row.get('image', ''))

            try:
                img = self.load_image(filename)
                ax.imshow(img)
                ax.axis('off')

                # Get positive findings
                positive_labels = [label for label in self.chexpert_labels if row[label] == 1]
                title = f"Sample {idx+1}\n" + "\n".join(positive_labels[:3])
                if len(positive_labels) > 3:
                    title += f"\n+{len(positive_labels)-3} more"
                ax.set_title(title, fontsize=10)
            except Exception as e:
                ax.text(0.5, 0.5, f"Error loading image:\n{str(e)}",
                       ha='center', va='center')
                ax.axis('off')

        plt.tight_layout()
        plt.savefig('dataset_samples.png', dpi=150, bbox_inches='tight')
        print("\n✓ Visualization saved as 'dataset_samples.png'")
        plt.show()

    def create_train_test_split(self, test_size=0.2, random_state=42):
        """Create train/test split"""
        from sklearn.model_selection import train_test_split

        if self.merged_df is None:
            raise ValueError("Process data first")

        train_df, test_df = train_test_split(
            self.merged_df,
            test_size=test_size,
            random_state=random_state,
            stratify=self.merged_df['No Finding']  # Stratify by normal/abnormal
        )

        print(f"\n✓ Train set: {len(train_df)} images")
        print(f"✓ Test set: {len(test_df)} images")

        return train_df, test_df

    def save_processed_data(self, output_path='processed_indiana_data.csv'):
        """Save processed dataset with labels"""
        if self.merged_df is None:
            raise ValueError("Process data first")

        self.merged_df.to_csv(output_path, index=False)
        print(f"\n✓ Processed data saved to: {output_path}")

    def get_dataset_info(self):
        """Print comprehensive dataset information"""
        if self.merged_df is None:
            raise ValueError("Process data first")

        print("\n" + "="*70)
        print("INDIANA DATASET INFORMATION")
        print("="*70)
        print(f"Total Images: {len(self.merged_df)}")
        print(f"Image Directory: {self.images_path}")
        print(f"\nDataFrame Shape: {self.merged_df.shape}")
        print(f"\nColumns: {list(self.merged_df.columns)}")
        print("\nLabel Statistics:")
        for label in self.chexpert_labels:
            if label in self.merged_df.columns:
                count = self.merged_df[label].sum()
                print(f"  {label}: {count}")
        print("="*70)


# ============================================
# USAGE EXAMPLE
# ============================================

if __name__ == "__main__":
    # ADJUST THIS PATH TO YOUR DATASET LOCATION
    DATASET_PATH = "./indiana_dataset"  # Change this to your actual path

    print("Indiana University CXR Dataset - Phase 1: Data Loading")
    print("="*70)

    # Initialize loader
    loader = IndianaDatasetLoader(DATASET_PATH)

    # Step 1: Load data
    print("\nStep 1: Loading CSV files...")
    projections_df, reports_df = loader.load_data()

    # Display sample rows
    print("\n--- Sample from Projections ---")
    print(projections_df.head(3))

    print("\n--- Sample from Reports ---")
    print(reports_df.head(3))

    # Step 2: Merge data
    print("\nStep 2: Merging projections and reports...")
    merged_df = loader.merge_data()

    # Step 3: Filter frontal images only
    print("\nStep 3: Filtering frontal images...")
    frontal_df = loader.filter_frontal_images()

    # Step 4: Extract labels from reports
    print("\nStep 4: Extracting CheXpert-style labels...")
    # Adjust 'findings' to match your actual column name in indiana_reports.csv
    labeled_df = loader.extract_labels_from_reports(report_column='findings')

    # Step 5: Dataset info
    loader.get_dataset_info()

    # Step 6: Create train/test split
    print("\nStep 5: Creating train/test split...")
    train_df, test_df = loader.create_train_test_split(test_size=0.2)

    # Step 7: Save processed data
    print("\nStep 6: Saving processed data...")
    loader.save_processed_data('indiana_processed.csv')

    # Step 8: Visualize samples (optional)
    print("\nStep 7: Visualizing sample images...")
    try:
        loader.visualize_sample(num_samples=3)
    except Exception as e:
        print(f"Visualization skipped: {e}")

    print("\n" + "="*70)
    print("✓ PHASE 1 COMPLETE: Data is loaded and ready for Phase 2 (Neural Module)")
    print("="*70)

    print("\nNext Steps:")
    print("1. Check 'indiana_processed.csv' for processed dataset")
    print("2. Verify label extraction is correct")
    print("3. Ready to proceed to Phase 2: Neural Network Training")
```
